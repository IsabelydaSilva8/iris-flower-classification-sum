```python
%pip install pandas
import pandas as pd
# DataFrame- função da biblioteca para constuir uma tabela. data = onde estão os dados, columns = nome das colunas
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
```

```python
iris_df['target'] = iris.target # adiciona a coluna target ao DataFrame onde fica o rótulo de cada amostra
iris_df.head() # mostra na tela os primeiros 5 linhas da tabela, se quiser ver mais linhas, pode colocar iris_df.head(10) para ver as 10 primeiras linhas
```

```python
#adiciona o nome das amostras na tabela
iris_df['target names'] = pd.Categorical.from_codes(iris.target, iris.target_names)
iris_df.head()
```

```python
%pip install matplotlib # intala a biblioteca Matplotlib
# %matplotlib inline- exibe os gráficos abaixo da célula 
%matplotlib inline 
# iris_df.plot.scatter- faz um gráfico de dispersão, 'sepal length (cm)' e 'sepal width (cm)' são os nomes das colunas do DataFrame, c='target' é a cor dos pontos no gráfico, colormap='viridis' é o mapa de cores utilizado
iris_df.plot.scatter('sepal length (cm)', 'sepal width (cm)', c='target', colormap='viridis')
```

```python
from sklearn.model_selection import train_test_split
%matplotlib inline
#X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
# separa os dados para treino e para teste, test_size=0.33 significa que 33% dos dados serão usados para teste e 67% para treino
#  random_state=42 garante o embaralhamento dos dados de forma consistente, para que os resultados possam ser reproduzidos
X = iris.data #pergunta
y = iris.target #resposta
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```
